Hierarchical Inheritance: Multiple child classes inherit from a single parent class

Multiple Inheritance: A child class inherits from more than one parent class.

Hybrid Inheritance: A mix of two or more types of inheritance (like multiple and hierarchical).

# hierarchical inheritance

```python
#this is a example of hierarchical inheritance
class schoolmember:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def tell(self):
        print(self.name,self.age, end=" ")
class student(schoolmember):
    def __init__(self,name,age,rollno,per):
        super().__init__(name,age)
        self.rollno=rollno
        self.per=per
    def tell(self):
        super().tell()
        print(self.rollno,self.per)
class teacher(schoolmember):
    def __init__(self,name,age,sal):
        super().__init__()
        self.sal=sal
    
    
```

```python
monika=student("monika",19,5,86)
```

```python
monika.tell()
```

Useful in multiple inheritance to ensure proper method resolution.

```python
class person:
    def __init__(self,name,age):
        print("person called")
        self.name=name
        self.age=age
    
class student:
    def __init__(self,name,age,rollno,per):
        print("student called")
        super().__init__(name,age)
        self.rollno=rollno
        self.per=per
class sciencestudent(student,person):
    def __init__(self,name,age,rollno,per,stream):
        print("sciencestudent called")
        super().__init__(name,age,rollno,per)
        self.stream=stream
    def tell(self):
        print(self.name,self.age,self.rollno,self.per,self.stream)
```

```python
monika.tell()
```

```python
class point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        x=self.x+other.x
        y=self.y+other.y
        p=point(x,y)
        return p
    def __repr__(self):
        return f"x={self.x},y={self.y}"
```

```python
p1=point(2,4)
```

```python
print(p1)
```

```python
p2=point(3,4)
```

```python
print(p2)
```

```python
p1+p2
```
