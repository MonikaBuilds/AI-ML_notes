# inheritance-Inheritance in Python is when a new class (child class) gets the features and behavior (methods) from an existing class (parent class)

```python
1.simple inheritance
2. multilevel inheritance
3. multiple inheritance
4. hirarchical inheritance
5. hybrid inheritance

objectclass
overriding--> repr , str
super()
```

```python
#simple inheritance
class A:
    pass
class B(A):
    pass
```

```python
#multilevel inheritance
class A:
    pass
class B(A):
    pass
class C(B):
    pass
```

```python
#multiple inheritance
class A:
    pass
class B:
    pass
class C(A,B):
    pass
```

```python
#hirarachical inheritance
class A:
    pass
class B(A):
    pass
class C(A):
    pass
```

```python
#hybrid inheritance (multiple + hirarchical)
class A:
    pass
class B(A):
    pass
class C(A):
    pass
class D(B,C):
    pass
```

# multilevel inheritance

```python
# real time scenario
# code reusability
# coding in java or in c++ in this language automatically parent class constructor will be called
# this is rule is not for python
# super method or we can call with the help of class name
class person:
    def __init__(self,age,name):
        self.name=name
        self.age=age
        print("person constructor called")
class emp(person):
    def __init__(self,age,name,ids,sal):
        person.__init__(self,age,name)
        self.ids=ids
        self.sal=sal
        print("emp constructor called ")
class manager(emp):
    def __init__(self,age,name,ids,sal,bonus):
        emp.__init__(self,age,name,ids,sal)
        self.bonus=bonus
        print("manager constructor called")
```

```python
class person:
    def __init__(self,age,name):
        self.name=name
        self.age=age
        print("person constructor called")
class emp(person):
    def __init__(self,age,name,ids,sal):
        super().__init__(age,name)
        self.ids=ids
        self.sal=sal
        print("emp constructor called ")
class manager(emp):
    def __init__(self,age,name,ids,sal,bonus):
        super().__init__(age,name,ids,sal)
        self.bonus=bonus
        print("manager constructor called")
    def __repr__(self):
        return f"manager bonus will be {self.bonus} manager age is {self.age} manager name is {self.name} manager salary is {self.sal} manager id is {self.ids}"
```

```python
obj2=manager(19,"monika",2,50000,2000)
```

```python
# method overriding
# how  super works
# how to call previous coes
# use of REPR
class person:
    def __init__(self,age,name):
        self.name=name
        self.age=age
        print("person constructor called")
    def __repr__(self):
        return f"age {self.age}, name {self.name}"
class emp(person):
    def __init__(self,age,name,ids,sal):
        super().__init__(age,name)
        self.ids=ids
        self.sal=sal
        print("emp constructor called ")
    def __repr__(self):
        str=super().__repr__()
        return f"{str},id{self.ids},sal {self.sal}"
class manager(emp):
    def __init__(self,age,name,ids,sal,bonus):
        super().__init__(age,name,ids,sal)
        self.bonus=bonus
        print("manager constructor called")
        
    def __repr__(self):
        string=super().__repr__()
        return f"{string}, bonus {self.bonus}"
```

```python
obj3=manager(19,"monika",2,50000,2000)
```
