```python
#sample variance
def svar(data):
    n=len(data)
    mean=sum(data)/n
    deviation=[(x-mean)**2 for x in data]
    variance=sum(deviation)/(n-1)
    return variance
```

```python
print(svar(data))
```

```python
np.var(data)       #population variance
```

```python
import statistics
statistics.variance(data)
```

```python
#population variance
def pvar(data):
    n=len(data)
    mean=sum(data)/n
    deviation=[(x-mean)**2 for x in data]
    variance=sum(deviation)/(n-1)
    return variance  
```

```python
print(pvar(data))
```

```python
statistics.pvariance(data)  #population variance
```

```python
#how to check standard deviation
sstd=svar(data)        #population deviation
print(np.sqrt(sstd))
```

```python
np.std(data)      #sample deviation
```

```python
statistics.stdev(data)     #sample deviation
```

```python
statistics.pstdev(data)   #population deviation
```
