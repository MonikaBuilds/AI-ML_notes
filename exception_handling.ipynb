# exception handling

Python Exception Handling handles errors that occur during the execution of a program. Exception handling allows to respond to the error, instead of crashing the running program. It enables you to catch and manage errors, making your code more robust and user-friendly.



# Error

1. static error/compile-time error/syntax error: This happens when you write something wrong in your code, like missing punctuation or using the wrong keyword. Python catches these mistakes before running the program

2. dynami error/run-time error/ exception: This happens while the program is running. The code is written correctly, but something unexpected happens, like dividing by zero or using a variable that doesn’t exist.

# exceptional handling keyword in python 

1. try: The try block is used to write code that might cause an error. If an error occurs, Python jumps to the except block instead of stopping the program.

2. except: The except block is used to handle errors that occur in the try block. Without this, Python would crash when an error happens.

3. else: The else block runs only if no errors occur in the try block. It is used for code that should execute when everything works fine.

4. raise: The raise keyword is used to manually trigger an error when a specific condition occurs.

5. finally: The finally block runs no matter what, whether there is an error or not. It is often used to clean up resources (like closing a file or database connection).

```python
try:
    a=int(input("enter first number"))
    b=int(input("enter second number"))
    c=a/b
    print("div is",c)
    d=a+b
    print("sum is",d)
except:
    print("b should not be 0")
```
