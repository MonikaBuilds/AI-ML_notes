```python
import numpy as np

# Define a structured data type: name (string), age (int)
dt = np.dtype([('name', 'S10'), ('age', 'i4')])

# Create a structured array with three records
a = np.array([('alice', 25), ('bob', 30), ('charlie', 35)], dtype=dt)
```
