```python
class emp:
    def __init__(self,a,b,c):
        self.age=a
        self.name=b
        self.salary=c
    def show(self):
         print(self.age,self.name,self.salary)
    
```

```python
monika=emp(19,"monika",2000)
```

```python
monika.age
```

```python
monika.name
```

```python
monika.show()
```

```python
class car:
    def __init__(self,a,b,c):
        self.top_speed=a
        self.fuel_capacity=b
        self.disk_break=c
    def show(self):
        print(f"top speed {self.top_speed} fuel_capacity {self.fuel_capacity} disk_break {self.disk_break} ")
    def drive(self,x):
        print(f"it can drive {x}")
```

```python
audi8=car(350,15,"yes")
```

```python
audi8.show()
```

```python
audi8.drive("very fast")
```
