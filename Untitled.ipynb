```python
import pandas as pd

df = pd.DataFrame({
    'A': [10, 20, 30, 40, 50],
    'B': [20, 30, 40, 50, 60],
    'C': [200, 300, 400, 500, 600]
})

threshold = 0.95
df = df[
    ((df >= df.quantile(1 - threshold)) & (df <= df.quantile(threshold))).all(axis=1)
]
print(df)
```

```python
def replace_outliers_with_mean(x):
    mean=x.mean()
    std=x.std()
    lower=mean - 3*std
    upper=mean +3*std
    x[(x < lower) | (x > upper)] = mean
    return x
df = df.apply( replace_outliers_with_mean)
```
