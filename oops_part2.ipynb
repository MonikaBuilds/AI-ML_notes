# Class Method: A method that takes the class itself as the first argument (cls). It can modify class-level attributes and is defined using @classmethod.

 # Static Method: A method that doesn't take self or cls as the first argument. It doesn't modify class or instance variables and is defined using @staticmethod.

Class Method: Works with the class.

Static Method: Does not rely on class or instance.

```python
class math:
    @staticmethod
    def add_nos(a,b):
        c=a+b
        return c
    @staticmethod
    def mul_nos(a,b):
        c=a*b
        return c
```

```python
math.add_nos(10,15)
```

```python
math.mul_nos(10,5)
```

```python
class Demo:
    def display():   #internally this method is being treated as a static method
        print("inside display")
        
Demo.display()
```

Instance Methods: The most common method type. Able to access data and properties unique to each instance(object)

Static Method (utility method) : Cannot access anything else in the class. Totally self-contained code.

Class Methods; Can access limited methods in the class. can modify class specific details.
