```python
import pandas as pd
import numpy as np
data={
    "energy source":["solar","wind","hydropower","geothermal","blomass","nuclear"],
    "energy consumption(mwh)":[1200,np.nan,2900,np.nan,2500,3200],
    "cost(Millions$)":[200,400,np.nan,150,250,np.nan]
}
energy_df = pd.DataFrame(data)
print("original energy data with missing values:")
energy_df

```

```python
energy_df["energy consumption(mwh)"].fillna(energy_df["energy consumption(mwh)"].mean(),inplace=True)
energy_df["cost(Millions$)"].fillna(energy_df["cost(Millions$)"].mean(),inplace=True)
print("\Data after inputing missing value with mean:")
energy_df
```

```python
#remove rows with any missing values
#dropna has a parameter axis=1 or 0....0 is bydefault 0 means row
#if i put 1 the whole column will be removed
#it is used mainely when 75% data is nan then we can remove the column

cleaned_df = energy_df.dropna()    

print("original energy data with misssing values:")
cleaned_df
```

```python
#removes columns with any NaN values.
cleaned_df = energy_df.dropna(axis=1)    

print("original energy data with misssing values:")
cleaned_df
```

```python
from sklearn.preprocessing import MinMaxScaler  

scaler = MinMaxScaler()
energy_df[["energy consumption(mwh)", "cost(Millions$)"]] = scaler.fit_transform(
    energy_df[["energy consumption(mwh)", "cost(Millions$)"]]  
)

print("\nData after normalization (Min-Max Scaling):")
print(energy_df)
```

# Encoding is the process of converting categorical (text) data into numeric form so that machine learning algorithms can understand and work with it.

```python
energy_encoded_df=pd.get_dummies(energy_df,columns=["energy source"])
print("\nData after one-Hot encoding categorical variable:")
energy_encoded_df
```
