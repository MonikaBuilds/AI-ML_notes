# concatenation of two dataframe

```python
#concatenation of two dataframe
one=pd.DataFrame({
    'Name':['monika','mansi','aryan','krishna'],
    'subject_id':['sub2','subj4','subj3','subj5'],
    'marks_scored':[89,98,47,54]},
    index=[1,2,3,4])
two=pd.DataFrame({
    'Name':['reshma','pooja','heena','sharan'],
    'subject_id':['sub2','subj4','subj3','subj5'],
    'marks_scored':[45,87,57,44]},
    index=[1,2,3,4])
```

# we can connect two dataframes in two ways
# 1.horizontal concat
# 2.vertical concat

```python
pd.concat([one,two])    #horizontal concatenation
```

```python
pd.concat([one,two],keys=["X","Y"])
```

```python
pd.concat([one,two],ignore_index=True)
```

```python
pd.concat([one,two],axis=1)      #vertival concatenate
```

# merge and joins

```python
#sample data
data1={'A':[1,2,3], 'B':[4,5,6]}
df1=pd.DataFrame(data1,index=['a','b','c'])

data2={'C':[10,20,30],'D':[40,50,60]}
df2=pd.DataFrame(data2,index=['a','c','d'])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
#sample data
data1={'A':[1,2,3], 'B':[4,5,6],'key':['a','b','c']}
df1=pd.DataFrame(data1)

data2={'C':[10,20,30],'D':[40,50,60],'key':['a','b','d']}
df2=pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2)
```

# pd.merge(left,right,how='inner',on=None,
# left_om=None,left_index=False,
# right_index=False,sort=True)

```python

left=pd.DataFrame({
    'Name':['monika','mansi','aryan','krishna'],
    'subject_id':['sub2','subj4','subj3','subj5'],
    'marks_scored':[89,98,47,54]},
    index=[1,2,3,4])
right=pd.DataFrame({
    'Name':['reshma','pooja','heena','sharan'],
    'subject_id':['sub2','subj4','subj3','subj5'],
    'marks_scored':[45,87,57,44]},
    index=[1,2,3,4])
```

```python
left
```

```python
right
```

```python
pd.merge(left,right,on="subject_id")
```

```python
pd.merge(left,right,on="subject_id",how="left")
```

```python
pd.merge(left,right,on="subject_id",how="right")
```
