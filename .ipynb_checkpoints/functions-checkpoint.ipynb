# call the function() function to calculate and return the factorial of the number of the number given by the user

```python
def factorial(n):
    while n>1:
        n=n-1
        print(n)
```

```python
factorial(5)
```

```python
def factorial(n):
    while n>1:
        print(n)
        n=n-1
```

```python
factorial(5)
```

```python
def factorial(n):
    f=1
    while n>1: #10>1...iteration..
        f=f*n #10*1=10
        n=n-1 #9
        
    return f
```

```python
factorial(5)
```

```python
def factorial(n):
    f=1
    while n>1: #10>1...iteration..
        f=f*n #10*1=10
        n=n-1 #9
    print(f)
```

```python
factorial(5)
```

```python
number=int(input("enter any number:"))
factorial(number)
```

```python
#Positional argument
def nos(a,b,c,d):
    return a*b*c*d
```

```python
nos(10) 
```

```python
nos(10,20,30,40)
```

```python
#Keyword arguments
def nos(a,b,c,d):
    return a*b*c*d
```

```python
mynums=nos(a=10,b=20,c=30,d=40)
```

```python
print(mynums)
```

```python
#Default arguments
def nos(a,b,d,c=0):
    return a*b*c*d
```

```python
nos(1,2,4)
```

```python
nos(1,2,d=4)
```
