what we are doing
we are going to preserve the info

this information can be a different different types(application or software)
information:
none(0)
debug(10) - Detailed info, usually for developers to debug problems.
info(20)- General info, everything is working as expected.
warning(30) - Something might be wrong, but program is still running fine
error(40) - A problem occurred, program couldn't do something.
critical(50) - A serious error. Program might crash or stop working correctly.
