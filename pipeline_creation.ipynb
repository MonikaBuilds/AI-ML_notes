```python
from sklearn.impute import SimpleImputer  #handling missing values
from sklearn.preprocessing import OneHotEncoder #handling categorical features
from sklearn.preprocessing import StandardScaler #feature scaling
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
dataset.columns
```

```python
dataset.info()
```

```python
categorical_column=['sex','smoker','day']
numeric_column=['total_bill','tip','size']
```

```python
dataset[categorical_column]
```

```python
dataset[numeric_column]
```

# create a pipeline


```python
numeric_pipeline = Pipeline(
    steps=[
        ('imputer',SimpleImputer(strategy="median")),  #handling missing value
        ('onehotencoder',StandardScaler())  #categorical features to numerical
    ]
)
cat_pipeline=Pipeline(
    steps=[
        ('imputer',SimpleImputer(strategy="most_frequent")),  #handling missing value
        ('onehot', OneHotEncoder(handle_unknown='ignore'))  #categorical features to numerical
    ]
)
```

```python
preprocessor=ColumnTransformer([
    ('numeric_pipeline',numeric_pipeline,numeric_column),
    ('cat_pipeline' , cat_pipeline,categorical_column)
])
```

```python
x_train=preprocessor.fit_transform(x_train)
```

```python
x_test=preprocessor.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
models={
    'Random Forest':RandomForestClassifier(),
    'Decision Tree':DecisionTreeClassifier(),
    'SVC' :SVC()
}
```

```python
print(list(models.keys())[0])
print(list(models.keys())[1])
print(list(models.keys())[2])
```

```python
list
```

```python
def evaluate_model(x_train,x_test,y_train,y_test,models):
    report={}
    for i in range(len(models)):
        model=list(models.values())[i]
        model.fit(x_train,y_train)

        y_pred=model.predict(x_test)
        acc_score = accuracy_score(y_test,y_pred)
        acc_score=round(acc_score*100,2)
        report[list(models.keys())[i]]=acc_score
    return report
```

```python
evaluate_model(x_train, x_test, y_train, y_test, models)
```
