```python
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stat
import numpy as np
import matplotlib.pyplot as plt  # Re-import to reset

def plot_data(sample):
    plt.figure(figsize=(10,8))
    plt.subplot(1,2,1)
    sns.histplot(sample, kde=True)
    plt.subplot(1,2,2)
    stat.probplot(sample, dist="norm", plot=plt)
    plt.show()
```

```python
#Q-Q plot-quantile quantile plot
plot_data(np.log(p))
```

# z-test

Use a Z-test when:

1. Sample size is large (typically n > 30).

2. Population standard deviation (σ) is known.

You want to compare:

1. A sample mean vs a population mean

2. Two sample means

3. A sample proportion vs population proportion

```python
np.mean(virat20_inning_score)
```

```python
#null hypothesis=both mean are same
```

```python
#two sample t test
#call function
stats.ttest_ind(virat20_inning_score,my20_inning_score)
```

```python
p_value=0.020526051623868606
```

```python

if p_value<0.05:
    print("you are going reject the null hypothesis means both sample mean are not same")
else:
    print("you are going accept the null hypothesis means both sample mean are same")
```
